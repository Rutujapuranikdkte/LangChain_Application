```python
# from langchain_community.document_loaders import PyPDFLoader #it is used to call the tool PyPDFLoader read pdf in langchain

# pdf_path = "DL3_Unit1.pdf"
# loader = PyPDFLoader(pdf_path)
# documents = loader.load() #converting each page into one document obj

# print("Pages:", len(documents)) #total number of pages
# print(documents[0].page_content[:300])

# 2. Load PDF (ALL pages)
from langchain_community.document_loaders import PyPDFLoader

pdf_path = list(uploaded.keys())[0]  # get uploaded file name 0-one file
loader = PyPDFLoader(pdf_path) #it is an object which can ready to load the file for reading
documents = loader.load() #converting each page into text uses pyPDFLoader(tool) ignoring images extracing readable text

print("Total pages:", len(documents))
```

```python
# from langchain_text_splitters import RecursiveCharacterTextSplitter ->import tool that breakes big text into small pieces

# text_splitter = RecursiveCharacterTextSplitter(
#     chunk_size=500, -piece contains max 500 characters
#     chunk_overlap=50 -take last 50 words from preious texts Avoid losing important info
# )

# chunks = text_splitter.split_documents(documents) #breaking doc into samll chunks

# print("Chunks:", len(chunks))
# print(chunks[0].page_content[:200])
# 3. Split text into chunks
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)
print("Total chunks:", len(chunks))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings #Import a tool called HuggingFaceEmbeddings coverting text intonumbers

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2" #embedding model-This is a pre-trained model from Hugging Face This model:Takes text,Converts it into numerical vectors
)

print("Working now")
```

```python
from langchain_community.vectorstores import FAISS #FAISS is a tool to store and search data quickly using vectors

vectorstore = FAISS.from_documents(chunks, embeddings) #text is stored in a smart searchable format

print("Vector DB created")
```

```python
retriever = vectorstore.as_retriever()
#vectorstore = where all your text (as vectors) is stored
#.as_retriever() = makes it usable for searching
print("Retriever ready")
```

```python
from transformers import pipeline #importing pipeline from transformer library to run AI code easily
from langchain_community.llms import HuggingFacePipeline #connects the Hugging Face model with LangChain

pipe = pipeline(
    "text2text-generation",   # important (not text-generation) input-text,generating output-text then best for question answering
     model="google/flan-t5-base", #pre-trained model by Google understand question with and generate ans
    max_length=512 #Limits for answer size
)

llm = HuggingFacePipeline(pipeline=pipe) #convert into a format LangChain understands

#HuggingFacePipeline wraps the transformers pipeline into a LangChain-compatible LLM interface so it can be used in chains like RetrievalQA.it is a wrapper (adapter) in LangChain
print("LLM loaded")
```

```python
from langchain.chains import RetrievalQA

qa_chain = RetrievalQA.from_chain_type( #finds relevant text from PDF
    llm=llm, #means passing your LLM model into the QA system
    retriever=retriever #searching relevant chunks from pdf
)

print("QA system ready")
```

```python
query = "What is the name of the topic?"
result = qa_chain.run(query) #The run() method executes the RetrievalQA chain by taking a query, retrieving relevant data, and generating an answer using the LLM.

print("Answer:", result)
```
